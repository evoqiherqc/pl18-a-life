## 1. The object

**PL18 (GUL Special Collections).** A Burmese palm-leaf *pothi* manuscript, held by Glasgow University Library, Special Collections.

Its covers are flat lacquered wooden boards, gilded with gold leaf and small gold dot decorations along the edges. Between the covers, a stack of cream palm-leaf folios is bound by silk cord passing through two binding holes pierced through every leaf. The script — incised into each leaf with a stylus and rubbed with black ink — runs across the leaves on both sides.

According to the catalogue, the text is the **Dhammapada**: 423 verses of Buddhist moral and spiritual instruction, attributed to the Buddha and one of the most widely read texts in Theravada Buddhism. The language is **Pali** and the script is **Burmese** — the standard convention for Pali texts produced in Burma.

A **GLASGOW UNIVERSITY** library label is visible on the upper red cover, with a manuscript number annotation. The label was added shortly after the manuscript was acquired by Glasgow University Library in 1873 (Weston handlist, acc. no. 202), having been gifted to a school in Edinburgh by a Burmese Buddhist monk in the 1830s. (Whether the monk physically came to Britain or whether the gift travelled via an intermediary is not specified in the catalogue. See [§6](#6.-Paradata-audit) for how I handled this gap in the film.)

The catalogue's identification of the text as the Dhammapada has not been verified by a specialist within living memory. As Robert Maclean (Special Collections) put it on the recording introducing this object:

> *"For a long time, [these manuscripts] have been inaccessible because we haven't really known what they are. We're out of circulation."*

## 2. The question

The brief asked for an object from Special Collections introduced to a public audience using AI tools, with a critical reflection on the human–AI collaboration.

The straightforward interpretation: make a polished AI-generated explainer, get a tick. That is the trap. AI image and video tools are now good enough that producing a slick four-minute heritage video is a couple of afternoons' work — and the result will look authoritative even though almost everything in it about an object like PL18 will be invented by a model that has never seen the manuscript and cannot read its script.

The animating question for this project is therefore: how do you use these tools honestly? The London Charter (2009) for the Computer-Based Visualisation of Cultural Heritage requires that interpretive content be clearly distinguished from documentary evidence, and that the methods of interpretation be made visible as paradata. AI tools, by default, do the opposite. They produce confident, plausible content with no audit trail.

The response is not to use less AI. It is to use it with the seams visible: to keep, in the public-facing work, a record of where the AI helped, where it had to be corrected, and where it produced confidently-wrong content that has been kept visible as evidence rather than fixed.

The film is the engagement piece. This notebook is the audit. The two halves argue, together, that human–AI collaboration in cultural heritage is most honest when the methodology is on the page, not behind it.

### What the 3D model is good for, and what it is not

## 5. Methodology

### 5.1 The visual bible

Before any image was generated, a written visual bible specified:

- **Aesthetic register:** painterly watercolour-and-ink, hand-touched, slightly imperfect
- **Six-colour palette** for use across the film, drawn loosely from the colour register of the manuscript itself:

| Hex | Colour |
|---|---|
| `#c9962b` | Burma gold |
| `#a85533` | Butter-lamp warm-red |
| `#87765c` | Drained warm |
| `#6b7b85` | Cool blue-grey |
| `#b09775` | Warm returning |
| `#d4a655` | Gold rhyme |

- **Camera language:** the camera is the manuscript. Low angles, close-ups on hands, slow drift only
- **Visual motifs:** silk cord, two binding holes, the changing hands across centuries
- **Aesthetic prohibitions:** no photoreal monks, no specific religious iconography, no exotic-Asia visual shorthand, no fast cuts, no on-screen text other than in the meta-ending

### 5.2 The style anchor

Every image-generation prompt began with the same opening paragraph, the **style anchor**:

> *Painterly stylised illustration in the spirit of Cartoon Saloon's* Wolfwalkers *and Studio Ghibli's* The Tale of the Princess Kaguya. *Hand-touched watercolour-and-ink aesthetic, visible brushstroke texture, slightly imperfect outlines, soft edges. Cinematic 16:9 framing. Warm earth-toned palette. Not photoreal. No text, no signage.*

The anchor was tested first on three diagnostic scenes — the manuscript at rest, a reading scene, and an archive scene — before any production generation began. All three returned coherent style across temperature and subject matter, which is what licensed committing to the pipeline.

### 5.3 The workflow

1. Visual bible written and locked
2. Style anchor drafted and tested on three diagnostic scenes
3. Three manuscript reference images generated and saved as Flow ingredients
4. 25 scene keyframes generated in Nano Banana, with reference images attached
5. First-pass review against (a) object fidelity, (b) style coherence, (c) audit of what the AI added unbidden
6. Targeted reshoots for shots with serious drift, using anti-drift prompt patterns
7. Image-to-video animation in Veo 3.1 Fast
8. Clip exports uploaded to a public YouTube playlist (Flow's full-export function failed; see [§6.10](#6.10-Film-delivery:-a-Flow-infrastructure-failure,-kept-visible))
9. Documentation — this notebook

## 6. Paradata audit

This is the central methodology of the project. Notable AI failures encountered during production are recorded below, with what was done about each, and (where applicable) what is being **kept visible as evidence rather than corrected**. This is the paradata layer the London Charter calls for, applied to generative AI.

### 6.1 Drift: Western book proportions

The most consistent failure mode of the image generation. The model knows the word *manuscript* but its representation of one is shaped by training data dominated by Western leather-bound books. Despite explicit prompts, several first-pass shots rendered PL18 as a thick rectangular Western book rather than as the long, slender, flat palm-leaf stack it actually is.

**Correction:** prompts were rewritten with a *CRITICAL OBJECT SHAPE* constraint block in capitals, separated typographically from the descriptive prose, emphasising the long-and-flat palm-leaf proportions rather than the thick-and-rectangular Western book the model defaulted to. The next regeneration produced correct proportions.

**Kept visible:** for Shot 4 (the binding), the first-pass Western-book rendering and the corrected palm-leaf rendering exist as a paired record. Only the corrected version appears in the final film; the methodological point is that the correction was needed, not that the result of it is invisible.

### 6.2 Drift: legible Burmese script

Several shots — the incising scene, the reading scene, the 3D scan — were prompted to show *"illegible flowing curves, Burmese-style script suggested but not legible,"* because the project explicitly does not claim to know what the manuscript says. The model produced what appears to be plausible Burmese script, possibly composed of real characters drawn from training data.

**Correction:** considered but not pursued. The shots are aesthetically strong, and the legibility issue is a meaningful finding for a project whose subject is exactly the AI's tendency to invent confident content where none should exist.

**Kept visible:** the script in several shots was generated by the AI and is not actual Pali Dhammapada text. The model's reluctance to produce intentionally illegible script — its tendency to fabricate plausible-looking content even when asked for unreadable abstraction — is a direct demonstration of the central concern of this project, kept on screen rather than fixed. The fabricated script is the visual equivalent of the fabricated shelfmark in §6.3: confident, plausible-looking, wrong.

### 6.3 Drift: invented catalogue number

The shot showing a librarian pasting the GLASGOW UNIVERSITY label onto the manuscript included a hand-written shelfmark reading **MS. 143/8**. This number is invented by the AI. The actual shelfmark is **PL18 (GUL Special Collections)**.

**Correction:** not pursued.

**Kept visible:** this is one of the most useful single moments in the project. A confident-looking but entirely fabricated archival reference number, generated in passing on a heritage object, is exactly the kind of risk the London Charter exists to flag. It is left in the film.

### 6.4 Drift: meta-ending setting

The first-pass generations of the film's meta-ending (shots 19, 20, 21) inherited so much warmth and wood-grain from the rest of the project that they continued to depict candlelit wooden tables and could not be read as "today."

**Correction:** the prompts for these shots were rewritten with an *ABSOLUTELY NO* negative-prompt block: no candles, no butter lamps, no wooden monastery furniture, no warm lamp light. The style register itself was nudged: less painterly, more graphic, slightly realistic. Reshoots produced clean modern desk scenes.

**Kept visible:** the candlelit drift stands as an example of how generative models default to whatever style register dominates their immediate context.

### 6.5 Drift: skin tone bleed

A reshoot of the Edinburgh school scene initially over-corrected and rendered the European hands as bluish-grey, "like ice creatures." The model had read *cool grey-blue palette* as a colour instruction applying globally rather than compositionally — including to living skin.

**Correction:** the prompt was rewritten to split palette guidance from skin-tone guidance: *"Palette: cool grey-blue dominates the room … but the hands themselves are normal warm pale skin tone … NOT blue or grey skin."*

**Kept visible:** the icy-hands version is preserved as a small but useful example of how the model interprets palette instructions globally rather than compositionally.

### 6.6 Drift: historical setting of the gift exchange

The first-pass version of Shot 8 (the moment of gifting) located the exchange in what appeared to be a European library interior. This was historically dubious: Burmese Buddhist monks did not generally travel to Britain in the 1830s, and the catalogue's brief account leaves the precise circumstances unclear (whether the monk physically came to Britain, or whether the gift travelled via an intermediary).

**Correction:** the prompt was rewritten to set the exchange explicitly inside a Burmese monastery interior, with deliberately ambiguous "pale visitor's hands" rather than a specific identifiable character.

**Kept visible:** *paradata note in the film:* the exchange is depicted in Burma rather than fabricating a transatlantic encounter for which there is no evidence. This is the kind of decision the London Charter calls *paradata*: declaring an interpretation rather than presenting it as documented fact.

### 6.7 Audio: the chant question

The original Veo 3.1 audio generation produced what it considered "Asian-religious-sounding" chant for the monastery scenes. The model has no way to distinguish Burmese Theravada Pali chant from Tibetan, Indian, Thai, or Chinese Buddhist traditions, and produced something that fused features of several — most likely shaped by what a Western ear is trained to expect "Buddhist chant" to sound like. For a project explicitly concerned with the AI's tendency to confabulate cultural content, this is exactly the kind of generalisation worth surfacing.

**Considered correction:** replace the AI chant with a CC-licensed Pali Dhammapada recitation from [sutta.org](http://sutta.org) or Access to Insight.

**Decision:** kept the AI-generated audio, with the generalisation visible. Substituting a real chant would have created a soundscape whose authenticity claim was uneven — visuals AI-generated and acknowledged, audio real but passing as AI. The decision to leave the AI audio is itself a methodology choice: it lets the generic chanting stand as audible evidence of the same training-distribution gravity that produces the European library interior in §6.6 and the fabricated shelfmark in §6.3, but in sound rather than image.

### 6.8 Drift: animation stiffness as a cost of image-first generation

The image-first pipeline (still keyframe in Nano Banana, then keyframe-conditioned Veo animation) preserves the visual bible's constraints but produces visibly stiffer motion than text-only Veo generation would have. This is a real trade-off rather than a hidden one.

**Correction:** considered but not pursued. The available correction was to shift selected shots to text-only Veo for fluid motion, accepting some constraint drift in exchange.

**Kept visible:** constraint preservation matters more for this project's claims than motion fluidity does — a fluid scene that drifts from the manuscript's actual proportions is worse for the project than a stiff scene that holds them. The cost is named here rather than concealed. There is no neutral methodology: every choice in a generative pipeline trades one virtue for another, and honest paradata records the trade rather than presenting the chosen option as if it had no cost.

### 6.9 The reversing horse-and-cart

In one Edinburgh-bound shot, the horse-drawn cart appears to reverse rather than advance. The honest cause is that the AI struggles with backwards movement: every attempt to prompt the cart into smooth forward motion produced something subtly worse than the previous attempt — a small "Father Ted dents his car" moment.

**Considered correction:** further regenerations.

**Decision:** kept it. Aside from the reversing motion, the rest of the shot reads cleanly, and there is something useful (and faintly funny) about a generative-AI methodology whose limits announce themselves as visibly as a cart pointing the wrong way. Generative video is finite-budget in prompts, not just in compute, and that constraint shapes which drifts the human chases and which they leave visible.

### 6.10 Film delivery: a Flow infrastructure failure, kept visible

The intended deliverable was a single rendered MP4 of the assembled film. Google Flow's full-export function consistently crashed the project's hardware when asked to render the complete sequence. Online discussion indicated that the problem was widely reported and not specific to this hardware, with the two commonly suggested workarounds being (a) screen-recording the in-Flow playback, and (b) downloading individual clips and reassembling externally.

**Correction:** option (a) would have produced a lower-quality copy of an output the project already possessed in higher quality. Option (b) would have required external editing software that hardware constraints made unreliable within the project's time budget.

**Kept visible:** the individual clip exports were uploaded to a public YouTube playlist organised in viewing order, embedded in the project site. The transitions between clips are AI infrastructure showing through rather than editorial choices. Contemporary generative AI is excellent at the headline output and surprisingly fragile at adjacent infrastructure: a model can synthesise minutes of painterly animation but cannot reliably hand the user a single file containing it. Consistent with the well-known cases in which the same systems flounder on tasks much simpler than their flagship capabilities — a structural feature of where generative AI is technically mature and where it is not. A small but instructive case of the technology evolving faster than the user-experience around it can keep up.

### 6.11 Two cross-cutting findings

Beyond the individual drifts, two patterns emerged across the production that are worth naming because they speak to the project's evaluative argument.

**Pattern A: the model defaults toward the centre of its training distribution.**
When asked for an unusual cultural object (a Burmese palm-leaf manuscript), the model consistently pulled toward the most-represented adjacent category (an ornate Western book). When asked for an unusual visual operation (illegible script that looks like script), it pulled toward the most-represented adjacent category (legible script). This is not exotic. It is exactly the behaviour to expect from a model trained on a distribution heavily weighted toward Western, Anglophone, modern visual material. **Naming the bias makes it auditable.**

**Pattern B: the model's most useful moments are its most embarrassing ones.**
The fabricated shelfmark, the confidently-rendered Burmese script, and the drift toward training-distribution defaults in early gift-exchange shots are all examples of content the model generated *because no one stopped it*. Each is a small, contained example of the central concern of the project. I have kept them. The labour of the human is not to make the AI's work invisible but to make it auditable. **The mistakes, kept and annotated, do more methodological work than the corrections.**

## 7. The Lovelace–Boden–Ridler test

The course's evaluation framework combines three approaches to evaluating AI creativity. The full discussion of each appears in the separate evaluation report. The summary below shows how the project's design maps onto the framework.

### 7.1 Lovelace 2.0 (Bringsjord, Bello, Ferrucci 2003; Riedl 2014)

The test asks whether output meets natural-language constraints **B**, set by a human evaluator, where constraints are non-trivial. **The key reframe**: it is the human–AI *collaboration* that is being evaluated, not the AI alone. Each AI generation does not need to pass; the methodology as a whole must.

**For this project:** the constraints were declared in advance (the visual bible, the aesthetic prohibitions, the wordlessness, the hands-not-faces device). Many individual generations did not, on their own, meet those constraints — see the drifts in §6. The human–AI collaboration meets them by virtue of the audit-and-correct loop documented above.

### 7.2 Boden's three criteria

A creative output should be **new**, **surprising**, and **valuable**.

- **New:** P-creative for the maker; not H-creatively original. The novelty being claimed is modest: an applied demonstration of what honest AI heritage interpretation might look like, not a new genre
- **Surprising:** **exploratory** rather than transformational, in Boden's typology. The surprise is in *where the AI fails* and what it fails toward — Western books, legible script, fabricated catalogue numbers — not in the polish of its successes
- **Valuable:** value is measured against (a) public engagement with a previously-inaccessible object, (b) modelling honest AI use for cultural heritage, (c) a worked demonstration of the London Charter's principles applied to generative AI

### 7.3 Ridler's critique

Anna Ridler argues that the Lovelace and Boden frameworks both pay too little attention to **intention, agency, labour, and medium** — and especially to *invisible labour*: the data curation, prompt iteration, selection, and post-processing that gets erased when results are presented as "AI-generated."

**For this project:** the drifts recorded in §6 are examples of invisible labour made visible.

- **Intention** is declared in the visual bible
- **Agency** is named: the human chose constraints, the model produced, the human selected and corrected
- **Labour** is logged in this notebook
- **Medium** is hybrid and explicitly multi-form: stills, video, 3D model, notebook, repository, evaluation report

The methodology and the evaluation framework were drafted alongside one another, which is why they map cleanly.
